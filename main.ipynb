```python
import fastf1
import pandas as pd
import os
```

```python
if not os.path.exists('f1_cache'):
    os.makedirs('f1_cache')
fastf1.Cache.enable_cache('f1_cache')
```

```python
def get_raw_race_results(year):
    print(f"--- Fetching Data for {year} ---")
    schedule = fastf1.get_event_schedule(year)
    
    all_race_data = []
    
    for index, event in schedule.iterrows():
        # Skip testing sessions, we only want real race weekends
        if event['EventFormat'] == 'testing':
            continue
            
        try:
            print(f"Loading {year} Round {event['RoundNumber']}: {event['EventName']}...")
            
            # Load only the Race session ('R'). 
            # We disable telemetry and weather to make the download very fast.
            session = fastf1.get_session(year, event['EventName'], 'R')
            session.load(telemetry=False, weather=False, messages=False)
            
            results = session.results
            
            # Loop through each driver's result in that race
            for _, driver in results.iterrows():
                all_race_data.append({
                    'Year': year,
                    'Round': event['RoundNumber'],
                    'Race_Name': event['EventName'],
                    'Driver_Code': driver['Abbreviation'],   # e.g., 'VER', 'NOR', 'HAM'
                    'Team': driver['TeamName'],              # e.g., 'Red Bull Racing'
                    'Finish_Position': driver['Position'],   # 1.0, 2.0, etc. (NaN if DNF)
                    'Points_Scored': driver['Points'],       # 25.0, 18.0, etc.
                    'Status': driver['Status']               # Tells you if it was an engine failure, crash, or finished
                })
        except Exception as e:
            # If a race hasn't happened yet (like late 2026 races), it will skip
            print(f"Skipping {event['EventName']} (Data not available yet)")
            
    return pd.DataFrame(all_race_data)
```

```python
df_agg_2025 = pd.merge(df_agg_2025, df_reliability, on="Driver_Code", how="inner")
```

```python
import numpy as np

# Copy your aggregated 20-row dataframe into the new binning dataframe
# Change 'your_merged_df' to the actual name of the dataframe from your screenshot
df_bin_2025 = df_agg_2025.copy()

# 1. Calculate the Z-Scores for all three columns
cols_to_zscore = ["Points_Scored", "Team_Point", "Finished"]

for col in cols_to_zscore:
    mean_val = df_bin_2025[col].mean()
    std_val = df_bin_2025[col].std()
    df_bin_2025[f"{col}_Z"] = (df_bin_2025[col] - mean_val) / std_val

# 2. Discretize Driver Skill
conditions_driver = [
    (df_bin_2025["Points_Scored_Z"] > 1.5),
    (df_bin_2025["Points_Scored_Z"] > 0.5),
    (df_bin_2025["Points_Scored_Z"] > -0.5)
]
choices_driver = ["Elite", "Good", "Mid"]
df_bin_2025["Driver_Skill_25"] = np.select(conditions_driver, choices_driver, default="Low")

# 3. Discretize Team Legacy
conditions_team = [
    (df_bin_2025["Team_Point_Z"] > 1.0),
    (df_bin_2025["Team_Point_Z"] > -0.5)
]
choices_team = ["Top_Tier", "Midfield"]
df_bin_2025["Team_Legacy_25"] = np.select(conditions_team, choices_team, default="Backmarker")

# 4. Discretize Reliability
conditions_rel = [
    (df_bin_2025["Finished_Z"] > -1.0) 
]
choices_rel = ["Solid"]
df_bin_2025["Reliability_26"] = np.select(conditions_rel, choices_rel, default="Fragile")

# View the final dataset for Netica
display_cols = ["Driver_Code", "Driver_Skill_25", "Team_Legacy_25", "Reliability_26"]
```

```python
df_parent = df_bin_2025[display_cols]
```

```python
df_pace = df_2026.dropna(subset=['Finish_Position']) 
df_pace_avg = df_pace.groupby('Driver_Code')['Finish_Position'].mean().reset_index()

# 2. Discretize into Netica States
conditions_pace = [
    (df_pace_avg['Finish_Position'] <= 3.5),
    (df_pace_avg['Finish_Position'] <= 10.5)
]
choices_pace = ['P1_P3', 'P4_P10']
df_pace_avg['Expected_Race_Pace_26'] = np.select(conditions_pace, choices_pace, default='Out_of_Points')

# 3. Merge this new column into your existing df_parent
df_final = pd.merge(df_parent, df_pace_avg[['Driver_Code', 'Expected_Race_Pace_26']], on='Driver_Code', how='left')

# 4. Handle drivers who might not have finished a race in 2026 yet
df_final['Expected_Race_Pace_26'] = df_final['Expected_Race_Pace_26'].fillna('Out_of_Points')

# View the final result
```
